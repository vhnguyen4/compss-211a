# HW2: Publish a Web Page with GitHub Pages

The full assignment notebook, CSV, and website starter have moved to the
[HW2 website template repository](https://github.com/macss-berkeley/hw2-hidden-berkeley-template).

1. Open that repository and choose **Use this template → Create a new repository** to create your own public copy.
2. Clone **your copy** with GitHub Desktop and open its folder in VS Code.
3. Open `hw2_hidden_berkeley.ipynb` there and follow its instructions.

Complete the notebook in your own repository and submit it through
[HW2 on bCourses](https://bcourses.berkeley.edu/courses/1556737/assignments/9131881).
