# Jupyter notebook basics: files, cells, extensions, and kernels

This notebook explains what a Jupyter notebook is before we begin coding in an empty one.

The central idea is:

> A notebook is a saved `.ipynb` file. A kernel is a separate process that runs code and remembers temporary state.

## What the `.ipynb` file saves

When you save this notebook, the file can contain:

- Markdown source;
- code source;
- cell order and metadata;
- execution counts;
- displayed outputs and errors.

It does **not** preserve a running Python process. Variables, imports, installed packages, and temporary Colab files must be recreated when the kernel restarts.

## VS Code, extensions, and the selected kernel

A local kernel is a separate Python process on your computer. A Colab kernel is a separate Python process on Google's computer. Neither kernel is stored inside the `.ipynb` file.

In VS Code, use the kernel picker in the upper-right corner. Choosing **Colab** changes where code runs; it does not move the local notebook file into Google Drive.

## Two cell types

### Markdown cells

Markdown cells contain headings, explanations, links, and other prose. Rendering a Markdown cell changes how it is displayed; it does not send anything to Python.

### Code cells

Code cells contain source code. Editing or saving a code cell does not execute it. Code reaches the selected kernel only when you run the cell.

## Save, Run, and Restart are different operations

```text
SAVE     writes notebook content to the .ipynb file
RUN      sends one or more code cells to the selected kernel
RESTART  clears the kernel's memory; it does not erase the notebook
```

Use the next cells to see the difference.

## Why top-to-bottom order matters

Notebook cells can be executed in a different order from the order shown on the page. That can leave the kernel remembering variables created by an earlier experiment or an older version of a cell.

Try to get into the habit of:

1. save the notebook;
2. restart the kernel;
3. Run All from top to bottom;
4. read the first error, if one occurs;
5. save the clean result.
